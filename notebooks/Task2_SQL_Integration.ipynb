```python
query = """
WITH product_sales AS (
    SELECT
        Category,
        "Product Name",
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Category, "Product Name"
),
ranked_products AS (
    SELECT
        Category,
        "Product Name",
        ROUND(Total_Sales, 2) AS Total_Sales,
        RANK() OVER (
            PARTITION BY Category
            ORDER BY Total_Sales DESC
        ) AS Product_Rank
    FROM product_sales
)
SELECT
    Category,
    "Product Name",
    Total_Sales
FROM ranked_products
WHERE Product_Rank = 1
ORDER BY Category;
"""

top_category_products = pd.read_sql(query, engine)

top_category_products
```

```python
query = """
WITH category_sales AS (
    SELECT
        Category,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Category
)
SELECT
    Category,
    ROUND(Total_Sales, 2) AS Total_Sales
FROM category_sales
ORDER BY Total_Sales DESC;
"""

cte_result = pd.read_sql(query, engine)

cte_result
```

```python
query = """
SELECT
    Category,
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROW_NUMBER() OVER (
        PARTITION BY Category
        ORDER BY SUM(Sales) DESC
    ) AS Product_Number
FROM sales
GROUP BY Category, "Product Name";
"""

row_number_result = pd.read_sql(query, engine)

row_number_result.head(15)
```

```python
query = """
SELECT
    Category,
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    RANK() OVER (
        PARTITION BY Category
        ORDER BY SUM(Sales) DESC
    ) AS Sales_Rank
FROM sales
GROUP BY Category, "Product Name";
"""

rank_result = pd.read_sql(query, engine)

rank_result.head(15)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Month
)
SELECT
    Month,
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(
        LAG(Total_Sales) OVER (ORDER BY Month), 2
    ) AS Previous_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

lag_result = pd.read_sql(query, engine)

lag_result.head(15)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Month
)
SELECT
    Month,
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(
        LEAD(Total_Sales) OVER (ORDER BY Month), 2
    ) AS Next_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

lead_result = pd.read_sql(query, engine)

lead_result.head(15)
```

```python
query = """
CREATE VIEW IF NOT EXISTS monthly_sales_view AS
SELECT
    strftime('%Y-%m', "Order Date") AS Month,
    ROUND(SUM(Sales), 2) AS Total_Sales
FROM sales
GROUP BY Month;
"""

with engine.connect() as connection:
    connection.exec_driver_sql(query)

print("View created successfully!")
```

```python
query = """
SELECT *
FROM monthly_sales_view
ORDER BY Month;
"""

view_result = pd.read_sql(query, engine)

view_result.head(10)
```
